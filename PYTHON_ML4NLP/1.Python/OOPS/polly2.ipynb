Need to design and payment methon like credit card/paypal/net banking and the payment process are also different. Need to implement via an abstract class. There will be a pay function in every payment method.

```python
# create and abstract cass. which will force the other class must have pay method.

from abc import ABC,abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass

# create the child class 1.Credit card payment and 2.UPI payment
class CreditcardPayment(Payment):
    def pay(self,amount):
        print(f"Payment via credit card and the amount is {amount}")

class UpiPayment(Payment):
    def pay(self,amount):
        print(f"Payment via UPI and the amount is {amount}")

class PaypalPayment(Payment):
    def deposit(self,amount):
        print(f"deposited amount is  amount is {amount}")



credit1=CreditcardPayment()
UPI1=UpiPayment()


```
