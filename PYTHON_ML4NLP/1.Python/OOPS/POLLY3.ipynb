Algorithmic Order Execution Engine

```python
## The Parent Blueprint:
from abc import ABC,abstractmethod
class ExchangeGateway:
    def __init__(self,exchange_name,order_history):
        self.order_history=[]

    @abstractmethod
    def execute_order(self,ticker, quantity, price, order_type):
        pass

class NasdaqGateway(ExchangeGateway):
    def execute_order(self,ticker, quantity, price, order_type):
        if quantity%100!=0:
            print("Invalid log: Quantity is not proper")
            return
        else:
            print(f"Order can be executed with {self.order_history}")

class NseGateway(ExchangeGateway):
    def execute_order(self,ticker, quantity, price, order_type):
        total_value=quantity
        if total_value > 20000000:
            # Step 3: Print a limit-breach error
            print(f"Error: Transaction value ₹{total_value:,.2f} exceeds the single order limit of ₹2,00,00,000!")
            # Step 4: Return early (stops execution)
            return False
            
        # If the check passes, proceed with the order placement
        print(f"NSE Order accepted for {quantity} units of {ticker} at ₹{price}.")
        return True

        
```
