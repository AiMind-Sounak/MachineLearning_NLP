```python
###### Define the class -- here we are creating the bankaccount system

class bankAcc:
    def __init__(self,name,age,balance=0):
        self.name=name
        self.age=age
        self.balance=balance
        print(f"The initial opening balance of is {self.balance}")


    def depost(self,amount):
        self.amount=amount
        print(f"the deposited amount is {self.amount}")
        self.balance=self.balance+amount
        print(f"The current balance is {self.balance}")

    def withdrawn(self,amount):
        if(amount>self.balance):
            print("Insufficient amount present")
        else:
            self.balance=self.balance-amount
    def get_balance(self):
        return self.balance

```

```python
acc_1=bankAcc("aa",40,2000)
```

```python
## deposite amont
acc_1.depost(500)
```

```python
## withdrawl
acc_1.withdrawn(2100)
```

```python
present_amount=acc_1.get_balance()
print(present_amount)
```
