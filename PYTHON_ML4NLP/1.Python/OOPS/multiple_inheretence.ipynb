## MULTIPLE INHERITENCE 
## Problem statement

"""
equirements:
1) Create a BankAcc class:It should initialize with a starting financial balance.It must contain a deposit(amount) method that adds money to the balance and prints the updated status.

2) Create a RewardsProgram class:It should initialize a customer loyalty wallet tracking points, starting at 0.It must contain an add_points(amount) method that calculates and awards points based on a percentage of money spent (e.g., 10% back).

3) Create a PremiumAccount class:This class must inherit from both BankAcc and RewardsProgram simultaneously.Its constructor should properly initialize the attributes of both parent classes.Implement a custom method named swipe_card(charge) that checks if the account has enough money for a transaction. If it does, deduct the money from the bank balance, award the corresponding rewards points to the user, and print the transaction receipts. Otherwise, print a decline message.
"""


```python
class BankAcc:

    def __init__(self,balance):
        self.balance=balance
        print(f"The initial balance is {self.balance}")
    
    def deposit(self,amount):
        self.balance+=amount
        print(f"The updated balance is {self.balance}")

class RewardsProgram:

    def __init__(self,point):
        self.point=0
    def add_points(self,amount):
        self.point=self.point+int(amount*0.1)
        print(f"Earned {self.point} reward points!")

## child class which takes inherit both the classes

class PremiumAccount(BankAcc,RewardsProgram):
    def __init__(self,balance,point):
        ## initiate both parents behavious explicitly
        BankAcc.__init__(self,balance)
        RewardsProgram.__init__(self,point)

    def swipe_card(self,charge):
        if(self.balance>=charge):
            self.balance=self.balance-charge
            self.add_points(charge) ## from reward method
            print(f"Spent ${charge}. New Balance: ${self.balance}")
        else:
            print("Declined!")




```
