""" 
Autonomous Delivery Drones.
1) The Navigation & Navigation Control Module (GpsTracker)
2)The Power & Energy Systems Module (BatteryPack)

Detailed Requirements1. 
1. The GpsTracker Class (Parent 1)Constructor (__init__): Must initialize with a string 
identifier drone_id and a tracking tuple coordinates representing (latitude, longitude).Method 
update_coordinates(new_lat, new_long):Updates the tracking tuple.Prints: "Drone {drone_id} updated position
 to ({new_lat}, {new_long})"
 2. The BatteryPack Class (Parent 2)Constructor (__init__): 
 Must initialize a property battery_level (integer representing percentage) with a default value of 100.
 Method consume_power(amount):Safely decreases the battery_level by the specified amount.The battery 
 cannot drop below 0. If it reaches 0, print: "Battery critically low! System shutting down."
 3. The AutonomousDrone Class (Child Class)Inheritance: Must inherit from both GpsTracker and 
 BatteryPack.Constructor (__init__):Accepts drone_id, initial_coordinates, and an optional 
 initial_battery (defaulting to 100).Critical Test: You must explicitly initialize both parent classes 
 inside this constructor so that no attributes are left uncreated or unassigned.Method 
 fly_to_destination(dest_lat, dest_long, power_cost):This method orchestrates actions from both parents.
 
 Rule 1: It must check if the drone has enough power to complete the trip. If battery_level is lower than 
 power_cost, print: "Flight aborted: Insufficient battery power." and do not move.
 Rule 2: If power is sufficient, it should drain the battery using the consume_power logic and update the 
 location using the update_coordinates logic.Rule 3: Print a success receipt containing the updated position 
and remaining battery life.

  
"""


  

```python
## Define both the super calss
## GPS TRACKER CLASS
class GpsTracker:
    def __init__(self,drone_id,latitude,longitude):
        self.drone_id=drone_id
        self.latitude=latitude
        self.longitude=longitude
    
    def update_coordinates(self,new_lat,new_long):
        self.latitude=new_lat
        self.longitude=new_long

## Battery PACK class
class BatteryPack:
    def __init__(self,battery_level=100):
        self.battery_level=battery_level

    def consume_power(self,amount):
        self.battery_level-=amount
        if(self.battery_level<0):
            print("Battery critically low! System shutting down.")

```

```python
## Child class
class AutonomousDrone(GpsTracker, BatteryPack):
    def __init__(self, drone_id, latitude, longitude, battery_level=100):
        # 1. Initialize the GpsTracker parent class
        GpsTracker.__init__(self, drone_id, latitude, longitude)
        
        # 2. Initialize the BatteryPack parent class
        BatteryPack.__init__(self, battery_level)

    def fly_to_destination(self,dest_lat, dest_long, power_cost):
        if(power_cost>self.battery_level):
            print("Flight aborted: Insufficient battery power.")
            return ## stop execution eaarly to protect drone
        self.consume_power(power_cost)
        self.update_coordinates(dest_lat,dest_long)
        print(f"The trip is successful with remaining baterry {self.battery_level} "
              f"and updated co-ordinate is {self.latitude}, {self.longitude}")
        
```

```python
## test case
delivery_drone = AutonomousDrone(drone_id="Alpha-9", latitude=22.57, longitude=88.36, battery_level=80)

## Move successsfully
print("--- Trip 1 ---")
# 2. Execute a valid flight path
delivery_drone.fly_to_destination(dest_lat=22.65, dest_long=88.43, power_cost=30)
```
