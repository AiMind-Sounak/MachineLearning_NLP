## Alerting and Notification Microservice.

```python
class NotificationChannel:
    def __init__(self,sender_id):
        self.sender_id=sender_id

    def send(self,recipient, message_body):
        raise NotImplementedError("")

class EmailChannel(NotificationChannel):
    def send(self,recipient, message_body):
        if "@" not in recipient:
            print(f"Invalid Email Address: {recipient}")
        else:
            print(f"Email sent from {self.sender_id} to {recipient}: [Subject: System Alert] {message_body}")
    

class SmsChannel(NotificationChannel):
    def send(self,recipient, message_body):
        if len(message_body)>160:
            print(f"SMS failed: Message exceeds character limit.")
        else:
            print(f"SMS text dispatched from {self.sender_id} to {recipient}: {message_body}")


def dispatch_alerts(channels_list, recipient, message):
    for channel in channels_list:
        
        channel.send(recipient, message)
```

```python
## test case preparation
# Instantiate the polymorphic channel processors
email_service = EmailChannel(sender_id="alerts@company.com")
sms_service = SmsChannel(sender_id="+15550199")

active_channels = [email_service, sms_service]

print("--- Test Case 1: Standard Alert ---")
short_msg = "Your servers are running smoothly."
dispatch_alerts(active_channels, recipient="admin@company.com", message=short_msg)

print("\n--- Test Case 2: Carrier Constraints Test ---")
long_msg = "System Warning: CPU utilization has crossed 92% on instance cluster-B. Please log in to the management console immediately to spin up redundant backup nodes."
dispatch_alerts(active_channels, recipient="admin@company.com", message=short_msg)
```
